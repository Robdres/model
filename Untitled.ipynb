```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import cm
import numpy as np

# ---------------------------------------
# Datos
# ---------------------------------------

data = [
    {"model": "LinearRegression", "r2_test": 0, "r2_train": 0.3463, "mae": 0.3941, "rmse": 0.5710, "train_time_sec": 2.0967},
    {"model": "Ridge", "r2_test": 0.3730, "r2_train": 0.3415, "mae": 0.3997, "rmse": 0.5747, "train_time_sec": 2.5198},
    {"model": "Lasso", "r2_test": 0, "r2_train": -0.3333, "mae": 0.6605, "rmse": 0.8127, "train_time_sec": 2.5154},
    {"model": "XGBRegressor", "r2_test": 0.9235, "r2_train": 1.0, "mae": 0.0782, "rmse": 0.1257, "train_time_sec": 171.8372},
    {"model": "HistGBR", "r2_test": 0.3757, "r2_train": 0.9847, "mae": 0.2185, "rmse": 0.3742, "train_time_sec": 55.9082},
    {"model": "MLPRegressor", "r2_test": 0.64, "r2_train": 0.9762, "mae": 0.201, "rmse": 0.3588, "train_time_sec": 209.8616},
    {"model": "RandomForest", "r2_test": 0.789, "r2_train": 0.9089, "mae": 0.2543, "rmse": 0.3879, "train_time_sec": 106.2525},
    {"model": "OrdinalRidge", "r2_test": 0, "r2_train": 0.2150, "mae": 0.3827, "rmse": 0.6186, "train_time_sec": 2.2544},
    {"model": "LogisticAT", "r2_test": 0, "r2_train": 0.3821, "mae": 0.2639, "rmse": 0.5137, "train_time_sec": 5.3587},
]

df = pd.DataFrame(data)

# ---------------------------------------
# Estilo general
# ---------------------------------------
plt.rcParams.update({
    "figure.figsize": (10, 5),
    "axes.titlesize": 18,
    "axes.labelsize": 14,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "axes.grid": True,
    "grid.alpha": 0.3,
    "grid.linestyle": "--"
})

# Paleta bonita
colors = cm.viridis(np.linspace(0.2, 0.85, len(df)))

# ---------------------------------------
# Función para hacer gráficos bonitos
# ---------------------------------------
def pretty_bar(x, y, title, ylabel):
    fig, ax = plt.subplots()
    bars = ax.bar(x, y, color=colors, edgecolor="black", alpha=0.9)

    # Nombre de modelos en diagonal
    plt.xticks(rotation=45, ha="right")

    ax.set_title(title)
    ax.set_ylabel(ylabel)

    # Borde del frame más limpio
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    # Etiquetas encima de las barras
    for bar in bars:
        h = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, h,
                f"{h:.2f}", ha="center", va="bottom", fontsize=9)

    plt.tight_layout()
    plt.savefig(ylabel+".png")
    plt.show()

# ---------------------------------------
# Gráficos finales
# ---------------------------------------
pretty_bar(df["model"], df["r2_test"], "R² en Test por Modelo", "R² Test")
pretty_bar(df["model"], df["mae"], "MAE por Modelo", "MAE")
pretty_bar(df["model"], df["rmse"], "RMSE por Modelo", "RMSE")
pretty_bar(df["model"], df["train_time_sec"], "Tiempo de Entrenamiento", "Segundos")
```
